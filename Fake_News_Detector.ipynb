# 3. Lemmatization Concepts

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

nltk.download('punkt_tab')
# Lemmatization which basically converts the words in its root form is necessary during text preprocessing for fitting the texts into models for proper evaluation of models 
```

```python
# Example

lemmatizer=WordNetLemmatizer()
input_str="been had done languages cities mice"

#Tokenize the sentence
input_str=nltk.word_tokenize(input_str)  # tokenize the string before lemmatization

#Lemmatize each word
for word in input_str:
    print(lemmatizer.lemmatize(word))
```

# Application of the ML Concepts in the Train and Test Dataset

```python
lemmatizer=WordNetLemmatizer()
for index,row in train.iterrows():
    filter_sentence = ''
    
    sentence = row['total']
    sentence = re.sub(r'[^\w\s]','',sentence) #cleaning
    
    words = nltk.word_tokenize(sentence) #tokenization
    
    words = [w for w in words if not w in stop_words]  #stopwords removal
    
    for word in words:
        filter_sentence = filter_sentence + ' ' + str(lemmatizer.lemmatize(word)).lower()
        
    train.loc[index,'total'] = filter_sentence
```

# Applying NLP Techniques

```python
# We are going to apply some NLP techniques to converts the text data into numbers for feeding into ML algorithms

# Bag of Words is an NLP technique in which, if we have a number of sentences, then total number of unique words will form in a dictionary with their frequency from the all sentences, and then each word will get a unique position out of the total number of unique position, a vector will form, and then 1 numbered binary represented vector will represent unique word with respective position in the dictionary

from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# CountVectorizer replaces the 1 in the vector by the frequency of the word
# The CountVectorizer converts text into numbers so that machine learning models can understand it.
```

 # Bag of Words / CountVectorizer

```python
# Example

corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]

vectorizer = CountVectorizer()            # CountVectorizer is assigned in an object named vectorizer 

X = vectorizer.fit_transform(corpus)      # fit() means learn all the unique words from the corpus, during transform() every sentence is converted into numbers Instead of words, each sentence becomes a vector.

print(vectorizer.get_feature_names_out()) # This prints the vocabulary, 9 unique words became the columns of the matrix
```

# TF-iDF Vectorizer

```python
# Tf-idf is a measure of originality of a word by comparing the number of times a word appears in a doc with the number of docs the word appear in.

# Example

def vectorize_text(features, max_features):
    vectorizer = TfidfVectorizer( stop_words='english',
                            decode_error='strict',
                            analyzer='word',
                            ngram_range=(1, 2),
                            max_features=max_features
                            #max_df=0.5 # Verwendet im ML-Kurs unter Preprocessing                   
                            )
    feature_vec = vectorizer.fit_transform(features)
    return feature_vec.toarray()
```

```python
tfidf_features = vectorize_text(['hello how are you doing','hi i am doing fine'],30)
```

```python
test_counts = count_vectorizer.transform(test['total'].values)
test_tfidf = tfidf.transform(test_counts)

#split in samples
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(tf_idf_matrix, Y_train, random_state=0)
```

# Pipeline

```python
#Assiging the variables again as once transformed vectors can't be transformed again using pipeline.
X_train = train['total']
Y_train = train['label']
```

```python
from sklearn.pipeline import Pipeline
import joblib
from sklearn import linear_model
from sklearn.feature_extraction.text import TfidfTransformer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
pipeline = Pipeline([
    ('vect', CountVectorizer()),
    ('tfidf', TfidfTransformer(norm='l2')),
    ('clf', linear_model.LogisticRegression(C=1e5)),
])
```

```python
pipeline.fit(X_train, Y_train)
```
